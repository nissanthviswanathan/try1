```python
import numpy as np 
import pandas as pd 

df =pd.read_csv("data/raw/nifty_5min.csv")
print(df.head())
print(df.info())
```

```python
df["date"] = pd.to_datetime(df["date"], format="%Y-%m-%d %H:%M:%S")
df = df.sort_values("date").reset_index(drop=True)
```

```python
print(df.isna().sum())
```

```python
df["trading_date"] = df["date"].dt.date
df["intraday_diff"] = df.groupby("trading_date")["date"].diff()
bad_gaps = df[
    df["intraday_diff"].notna() &
    (df["intraday_diff"] != pd.Timedelta(minutes=5))
]

print(bad_gaps[["date", "intraday_diff"]])
```
